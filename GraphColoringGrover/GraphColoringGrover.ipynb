## Using Grover's algorithm to solve graph coloring problems

Finally, now that you've implemented the oracles for two different kinds of graph coloring problems, you can experiment with using Grover's algorithm to solve them.

For graph coloring problems, it's not as easy to know the number of solutions to the problem upfront as it was for the prefix function used in the "Grover's Search Algorithm" kata.
Experiment with choosing the number of iterations at random. How does this affect the success probability?

> Notice that the example used in this demo, a three-vertex graph with two edges $0 \leftrightarrow 1 \leftrightarrow 2$, has the search space of size $4^3 = 64$.
> A bit over half of the colorings in the search space are valid vertex colorings. You can count them by assigning any color to vertex $1$ and then assigning any of the remaining three colors to vertices $0$ and $2$ independently, to get $4 \cdot 3 \cdot 3 = 36$ solutions to the problem.
> This example is one of the scenarios in which Grover's search behaves in a non-intuitive way: not doing any iterations 
> yields a success probability of over $50\%$, so the first iteration actually reduces the probability of getting a correct answer!

```python
from functools import partial
from psiqdk.workbench import QPU, Qubits

def int_to_bool_array(n: int, num_bits: int) -> list[bool]:
    """
    Converts an integer to a list of Booleans representing its bits.
    The order is from the least significant bit to the most significant bit.
    """
    return [bool(n & (1 << ind)) for ind in range(num_bits)]


def grovers_search(qpu: QPU, n_qubits: int, marking_oracle: callable, n_iter: int) -> int:    
    reg = Qubits(n_qubits, "reg", qpu)
    minus = Qubits(1, "minus", qpu)

    # Initialize the minus qubit.
    minus.x()
    minus.had()

    # Prepare the system in the state |all⟩.
    reg.had()

    # Do Grover's iterations.
    for _ in range(n_iter):
        # Apply the phase oracle = the marking oracle via phase kickback.
        marking_oracle(reg, minus)

        # Apply "reflection about the mean".
        reg.had()
        # Apply conditional phase flip up to a global phase (since it doesn't affect measurement outcomes).
        (~reg).reflect()
        reg.had()

    # Measure each qubit individually to get the result.
    return [reg[ind].read() for ind in range(n_qubits)]


def oracle_vertex_coloring(V: int, edges: list[(int,int)], x: Qubits, y: Qubits) -> None:
    '''Oracle for checking vertex coloring (problem 4).'''

    def oracle_color_equality(x0 : Qubits, x1 : Qubits, y : Qubits) -> None:
        for i in range(len(x0)):
            x1[i].x(cond = x0[i])
        
        y.x(cond=x1 == 0)

        for i in range(len(x0)):
            x1[i].x(cond = x0[i])

    edgesNumber = len(edges)
    if edgesNumber == 0:
        y.x()
        return
    conflicts = Qubits(edgesNumber, "conflicts", x.qpu)
    for i in range(edgesNumber):
        v0, v1 = edges[i]
        oracle_color_equality(x[2 * v0 : 2 * (v0 + 1)], x[2 * v1 : 2 * (v1 + 1)], conflicts[i])
    y.x(cond=conflicts == 0)
    for i in range(edgesNumber):
        v0, v1 = edges[i]
        oracle_color_equality(x[2 * v0 : 2 * (v0 + 1)], x[2 * v1 : 2 * (v1 + 1)], conflicts[i])
    conflicts.release()

def is_vertex_coloring_valid(V: int, edges: list[tuple[int, int]], colors: list[int]) -> bool:
    '''Classical function for checking vertex coloring (problem 1).'''
    for (v0, v1) in edges:
        if colors[v0] == colors[v1]:
            return False
    return True


# Experiment with the parameters to explore algorithm behavior in different conditions!
# The 0 -- 1 -- 2 graph from the examples used in the problems
V = 3
edges = [(0, 1), (1, 2)]

num_qubits = 2 * V + 1 + len(edges)
qpu = QPU()

marking_oracle = partial(oracle_vertex_coloring, V, edges)

for iterations in range(10):
    success = 0
    for _ in range(100):
        qpu.reset(num_qubits)
        bits = grovers_search(qpu, 2 * V, marking_oracle, iterations)
        
        # Convert measurement results to a coloring - array of integers
        colors = [bits[2 * i] + bits[2 * i + 1] * 2 for i in range(V)]
        
        if is_vertex_coloring_valid(V, edges, colors):
           success += 1

    print(f"{iterations} iterations - {success}% success rate")
```
