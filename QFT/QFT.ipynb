## Quantum Fourier transform: the definition

*Discrete Fourier transform* (DFT) is a classical transformation that acts on vectors of complex numbers of fixed length $N$. The effect of DFT on a vector $x_0, x_1, ..., x_{N-1}$ is another vector $y_0, y_1, ..., y_{N-1}$ defined as follows:

$$y_k = \frac1{\sqrt{N}}\sum_{j=0}^{N-1} x_j e^{2\pi i jk/N}$$

Quantum Fourier transform is the quantum version of the DFT. It acts on $n$-qubit quantum states and is defined via its effects on basis states as follows:

$$\ket{j} \rightarrow \frac1{\sqrt{2^n}}\sum_{k=0}^{2^n-1}  e^{2\pi i jk/2^n} \ket{k}$$

The result of applying QFT to an arbitrary superposition state with amplitudes $x_j$ can thus be expressed as a superposition state with amplitudes $y_k$ that are exactly the DFT of the amplitudes $x_j$ for $N = 2^n$:

$$\sum_{j=0}^{N-1} x_j\ket{j} \rightarrow \sum_{k=0}^{N-1} y_k\ket{k}$$

In the first part of this kata, you will learn to implement quantum Fourier transform using the iterative algorithm.

> All the numbers used in this kata use little-endian encoding: the *least* significant bit of the number is stored in the *first* (leftmost) bit/qubit.
>
> This means that you represent an integer $j$ as a binary bit string in the following format:
>
> $$j = \overline{j_{n-1}j_{n-2} ... j_1j_0} = j_0 2^0 + j_1 2^1 + ... j_{n-1} 2^{n-1}$$
>
> You can also use this notation for binary fractions: 
> 
> $$\overline{0.j_{n-1} j_{n-2} ... j_0} = \frac{j_{n-1}}{2^1} + \frac{j_{n-2}}{2^2} + ... + \frac{j_0}{2^n}$$

## Single-qubit QFT

Let's start by considering quantum Fourier transform for one qubit, that is, $n = 1$. 

In this case, $N = 2^n = 2$, and the effect of the QFT on the basis states is defined as 

$$\ket{j} \rightarrow \frac1{\sqrt{2}}\sum_{k=0}^{1}  e^{2\pi i jk/2} \ket{k} = \frac1{\sqrt2} \big( \ket{0} + e^{2\pi i \cdot \frac{j}{2}} \ket{1} \big)$$

Since there are only two basis states, we can spell out the effect of the QFT on them explicitly:

$$\textrm{QFT} \ket{0} = \tfrac1{\sqrt2} \big( \ket{0} + e^{2\pi i \cdot \frac{0}{2}} \ket{1} \big) = \tfrac1{\sqrt2} \big( \ket{0} + \ket{1} \big)$$

$$\textrm{QFT} \ket{1} = \tfrac1{\sqrt2} \big( \ket{0} + e^{2\pi i \cdot \frac{1}{2}} \ket{1} \big) = \tfrac1{\sqrt2} \big( \ket{0} - \ket{1} \big)$$

We can now define the effect of the QFT on arbitrary superposition states:

$$\textrm{QFT} \big( x_0 \ket{0} + x_1 \ket{1} \big) = x_0 \tfrac1{\sqrt2} \big( \ket{0} + \ket{1} \big) + x_1 \tfrac1{\sqrt2} \big( \ket{0} - \ket{1} \big) = \tfrac1{\sqrt2} \big((x_0 + x_1) \ket{0} + (x_0 - x_1) \ket{1}\big)$$

## The equivalence of two QFT notations

How to show that the transformation defined in the last exercise is the same as the original definition of QFT?
In other words, how to prove the equality below?

$$\frac1{\sqrt{2^n}} \sum_{k=0}^{2^n-1} e^{2\pi i \cdot \frac{jk}{2^{n}}} \ket{k} = $$

$$\begin{align*}
= & \tfrac1{\sqrt2} \big(\ket{0} + e^{2\pi i \cdot \overline{0.j_{n-1} j_{n-2} ... j_0}} \ket{1} \big) \otimes \\
\otimes & \tfrac1{\sqrt2} \big(\ket{0} + e^{2\pi i \cdot \overline{0.j_{n-2} ... j_0}} \ket{1} \big) \otimes ... \\
\otimes & \tfrac1{\sqrt2} \big(\ket{0} + e^{2\pi i \cdot \overline{0.j_1 j_0}} \ket{1} \big) \otimes \\
\otimes & \tfrac1{\sqrt2} \big(\ket{0} + e^{2\pi i \cdot \overline{0.j_0}} \ket{1} \big)
\end{align*}$$

The tensor product state (the second state in the equation above) is defined in terms of the digits $j_0, j_1, ..., j_{n-1}$.
Let's spell the exponents used in each term of the tensor product in terms of the integer $j$ instead:

$$e^{2\pi i \cdot \overline{0.j_{n-1} j_{n-2} ... j_0}} = e^{2\pi i \cdot \frac{j}{2^n}}$$

For rewriting the second term, recall that since $j_{n-1}$ can only be $0$ or $1$, $e^{2\pi i \cdot j_{n-1}}$ is either $e^0$ or $e^{2\pi i}$, which are both $1$. Any exponents of integer multiples of $e^{2\pi i}$ are $1$ and thus can be ignored. With this in mind, we can write

$$e^{2\pi i \cdot \overline{0.j_{n-2} ... j_0}} = e^{2\pi i \cdot j_{n-1}} e^{2\pi i \cdot \overline{0.j_{n-2} ... j_0}} = e^{2\pi i \cdot \overline{j_{n-1}.j_{n-2} ... j_0}} = e^{2\pi i \cdot \frac{2j}{2^n}}$$

And so on, until the last term:

$$e^{2\pi i \cdot \overline{0.j_0}} = e^{2\pi i \cdot \frac{j \cdot 2^{n-1}}{2^n}}$$

With those substitutions, the tensor product state looks as follows:

$$\begin{align*}
& \tfrac1{\sqrt2} \big(\ket{0} + e^{2\pi i \cdot \frac{j}{2^n}} \ket{1} \big) \otimes \\
\otimes & \tfrac1{\sqrt2} \big(\ket{0} + e^{2\pi i \cdot \frac{2j}{2^n}} \ket{1} \big) \otimes ... \\
\otimes & \tfrac1{\sqrt2} \big(\ket{0} + e^{2\pi i \cdot \frac{2^{n-2}j}{2^n}} \ket{1} \big) \otimes \\
\otimes & \tfrac1{\sqrt2} \big(\ket{0} + e^{2\pi i \cdot \frac{2^{n-1}j}{2^n}} \ket{1} \big)
\end{align*}$$

What happens if you open the brackets and compare the results to the original definition of the QFT?
Let's do this exercise for $n = 2$.

$$\tfrac1{\sqrt2} \big(\ket{0} + e^{2\pi i \cdot \frac{j}{4}} \ket{1} \big) \otimes
\tfrac1{\sqrt2} \big(\ket{0} + e^{2\pi i \cdot \frac{2j}{4}} \ket{1} \big) = \\
= \tfrac12 \big(\ket{00} + e^{2\pi i \cdot \frac{j}{4}} \ket{10} + e^{2\pi i \cdot \frac{2j}{4}} \ket{01} + e^{2\pi i \cdot \frac{3j}{4}} \ket{11} \big)$$

Remember that this kata uses little-endian encoding to map bit strings to integers. With that, the state can be rewritten using integers to represent basis states as follows:

$$\tfrac12 \big(\ket{0} + e^{2\pi i \cdot \frac{j}{4}} \ket{1} + e^{2\pi i \cdot \frac{2j}{4}} \ket{2} + e^{2\pi i \cdot \frac{3j}{4}} \ket{3} \big)$$

And that's exactly the original definition of QFT:

$$\frac1{\sqrt{4}} \sum_{k=0}^{3} e^{2\pi i \cdot \frac{jk}{4}} \ket{k}$$

The same equivalence can be demonstrated for an arbitrary $n$.
