## Canonical satisfiability problem

The Boolean satisfiability problem is the problem of determining whether there exists an assignment of Boolean variables $x_j$ for which the given Boolean formula evaluates to true.

The canonical representation for SAT problems is based on the conjunctive normal form of Boolean formulas.
To define this form, we'll need several definitions:

- The _variables_ are the Boolean variables $x_0, ..., x_{n-1}$ that act as inputs to the formula.
- A _literal_ is either a variable $x_j$ (a positive literal) or the negation of a variable $\neg x_j$ (called a negative literal).
- A _clause_ is a disjunction (an OR operation) of one or several literals, for example, $x_0 \vee \neg x_1$.
  Generally, 
  $$clause(x) = \bigvee_k y_{k},\text{ where }y_{k} =\text{ either }x_j\text{ or }\neg x_j\text{ for some }j \in \{0, \dots, n-1\}$$
- Finally, the _conjunctive normal form_ of a formula is a conjunction (an AND operation) of several clauses:
  $$f(x) = \bigwedge_i \big(\bigvee_k y_{ik} \big),\text{ where }y_{ik} =\text{ either }x_j\text{ or }\neg x_j\text{ for some }j \in \{0, \dots, n-1\}$$

In this kata, you will learn to implement marking oracles for SAT problems given as their conjunctive normal form descriptions.

## Exactly-1 3-SAT problem

Exactly-1 3-SAT problem (also known as "one-in-three 3-SAT") is a variant of a general SAT problem.
The structure of the formula that describes an instance of an exactly-1 3-SAT problem is exactly the same as that of the canonical SAT problem, with each clause consisting of exactly three literals.
However, the problem is to find an assignment of the variables that makes each clause have *exactly one* true literal, 
while in a normal SAT problem each clause can have one or more true literal to satisfy the formula.

Formally, the clauses of exactly-1 3-SAT problem can be defined via the use of a ternary operator that is `true` if and only if exactly one of the arguments is `true`. However, this kata uses the same formula notation for these clauses as the canonical SAT problem.

## Using Grover's algorithm to solve SAT problems

Now that you've implemented the oracles for two different kinds of SAT problems, you can experiment with using Grover's search algorithm to solve SAT problems.

Notice that in this case, the number of solutions to the problem is not as obvious as it was for the prefix function used in the "Grover's Search Algorithm" kata.
Experiment with choosing the number of iterations at random. How does this affect the success probability?

```python
from functools import partial
from psiqdk.workbench import QPU, Qubits

def grovers_search(n_qubits: int, marking_oracle: callable, n_iter: int):
    qpu = QPU(num_qubits=2 * n_qubits + 1)
    reg = Qubits(n_qubits, "reg", qpu)
    minus = Qubits(1, "minus", qpu)

    # Initialize the minus qubit.
    minus.x()
    minus.had()

    # Prepare the system in the state |all⟩.
    reg.had()

    # Do Grover's iterations.
    for _ in range(n_iter):
        # Apply the phase oracle = the marking oracle via phase kickback.
        marking_oracle(reg, minus)

        # Apply "reflection about the mean".
        reg.had()
        # Apply conditional phase flip up to a global phase (since it doesn't affect measurement outcomes).
        (~reg).reflect()
        reg.had()

    # Measure each qubit individually to get the result.
    return [reg[ind].read() for ind in range(n_qubits)]


def oracle_sat_formula(x: Qubits, y: Qubits, clauses: list[list[tuple[int, bool]]]) -> None:
    def oracle_sat_clause(x: Qubits, y: Qubits, literals: list[tuple[int, bool]]) -> None:
        ind, pos = literals[0]
        controls = x[ind] if pos else ~x[ind]
        for ind, pos in literals[1:]:
            controls |= x[ind] if pos else ~x[ind]

        y.x(cond=~controls)
        y.x()

    num_clauses = len(clauses)
    clause_ok = Qubits(num_clauses, "clause_ok", x.qpu)
    for ind in range(num_clauses):
        oracle_sat_clause(x, clause_ok[ind], clauses[ind])
    y.x(cond=clause_ok)
    for ind in range(num_clauses):
        oracle_sat_clause(x, clause_ok[ind], clauses[ind])
    clause_ok.release()


def f_sat_formula(args: list[bool], clauses: list[list[tuple[int, bool]]]) -> bool:
    def f_sat_clause(args: list[bool], literals: list[tuple[int, bool]]) -> bool:
        for (ind, pos) in literals:
            if pos and args[ind] or not pos and not args[ind]:
                return True
        return False

    for clause in clauses:
        if not f_sat_clause(args, clause):
            return False
    return True


# Experiment with the parameters to explore algorithm behavior in different conditions!
n, clauses = 2, [[(0, True)], [(1, True)]]  # 1 solution

marking_oracle = partial(oracle_sat_formula, clauses=clauses)
for n_iter in range(9):
    success = 0
    for _ in range(100):
        res = grovers_search(n, marking_oracle, n_iter)
        if f_sat_formula(res, clauses):
            success += 1
    print(f"{n_iter} iterations - {success}% success rate")
```
