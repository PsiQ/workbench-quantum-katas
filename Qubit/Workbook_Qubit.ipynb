# Problem 1. Get the amplitudes of a quantum state

The `pull_state` method can access the amplitudes of the state easily. 

```python
from psiqdk.workbench import Qubits

def fetch_single_qubit_amplitudes(reg: Qubits) -> list[complex]:
    state_vector = reg.pull_state()
    return state_vector
```
