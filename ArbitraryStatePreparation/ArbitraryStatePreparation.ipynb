### Demo: Testing simple state preparation routines

Now that we have implemented a naive state preparation algorithm, let's pause and consider how we can test our implementation. The katas test your solutions for you, of course, but once you're writing your own code, testing it becomes critically important.

To write tests that validate the quantum state of the program automatically, we need a way to extract that state and make it available to the rest of the code. Workbench provides a variety of methods to get the state vector of the system (see the [Testing and Debugging tutorial](https://construct.psiquantum.com/docs/psiqdk-workbench/new-tutorials/Testing-Debugging.html)), and we'll use one of them for testing our state preparation routine. Since we aim to prepare the state precisely, without any global phase discrepancy, we'll use the `pull_state` method of the QPU object, which returns the quantum state of the simulation as is.

Writing a test for a state preparation routine involves the following steps:

1. Select a set of test cases, that is, states that your code has to prepare correctly. You'll want to include a variety of test cases of different sizes, starting with single-qubit states. In our case, the state preparation code is supposed to handle only real amplitudes, so you don't need to include test cases with complex amplitudes, but you'll want to have test cases with a mix of positive and negative amplitudes, as well as corner cases such as basis states and basis states with a minus sign.

2. Run the state preparation code on each input individually, allocating a fresh qubit register each time. Since we're looking at preparing arbitrary superposition states, you can use Workbench [state vector simulator](https://construct.psiquantum.com/docs/psiqdk-workbench/new-tutorials/Simulating-WB-Programs.html#state-vector-simulator) (filter `>>state-vector-sim>>`, included in the default QPU setup).

3. Fetch the state of the register after the preparation routine and compare it to the state you wanted to prepare. The states should match!

The following code shows a simple testing harness that you can use to test `NaiveStatePrep` Qubrick from the previous problem.

```python
from math import atan2, sqrt
from psiqdk.workbench import QPU, Qubits, Qubrick, units
from pytest import approx

class NaiveStatePrep(Qubrick):
    def _compute(self, reg: Qubits, a: list[float], ctrl: Qubits | None = None) -> None:
        """Prepare the register reg in the superposition of all basis states defined by amplitudes a."""
        n = len(reg)
        if n == 1:
            theta = 2 * atan2(a[1], a[0])
            reg.ry(theta * units.rad, cond=ctrl)
        else:
            a_msb0 = a[:2 ** (n - 1)]
            a_msb1 = a[2 ** (n - 1):]
            b0 = sqrt(sum(aj * aj for aj in a_msb0))
            b1 = sqrt(sum(aj * aj for aj in a_msb1))
            
            self._compute(reg[-1], [b0, b1], ctrl)
            self._compute(reg[:-1], a_msb0, ctrl=ctrl | ~reg[-1])
            self._compute(reg[:-1], a_msb1, ctrl=ctrl | reg[-1])


def run_test_state_prep(qubrick: type[Qubrick], num_qubits: int, amps: list[float]) -> None:
    # Create QPU and allocate a fresh register.
    # If your solution uses auxiliary qubits, you'll need to adjust the num_qubits argument accordingly
    qpu = QPU(num_qubits=num_qubits)
    reg = Qubits(num_qubits, 'reg', qpu)
    # Instantiate the state prep Qubrick
    state_prep = qubrick()
    # Prepare the state
    state_prep.compute(reg, amps)
    # Get the state vector
    actual_amps = qpu.pull_state()
    # Check that the actual prepared state matches the expected one
    assert actual_amps == approx(amps)


# Run the test for the Qubrick on a variety of inputs and input sizes
for (n, amps) in [
    (1, [1.0, 0.0]),
    (1, [0.0, -1.0]),
    (1, [-0.6, 0.8]),
    (2, [-1., 0., 0., 0.]),
    (2, [1/sqrt(2), 0., 0., 1/sqrt(2)]),
    (2, [0.36, -0.48, 0.64, -0.48]),
    # More test cases omitted for brevity of the example
]:
    run_test_state_prep(NaiveStatePrep, n, amps)
print("Tests passed!")
```
